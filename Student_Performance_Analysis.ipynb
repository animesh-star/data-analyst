```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
df = pd.read_csv("student-mat.csv", sep=";")

df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["G3"])

plt.title("Boxplot of Final Mathematics Grade")
plt.xlabel("Final Grade (G3)")

plt.show()
```

```python
plt.figure(figsize=(9,6))

sns.boxplot(x="studytime", y="G3", data=df)

plt.title("Study Time vs Final Mathematics Grade")
plt.xlabel("Weekly Study Time")
plt.ylabel("Final Grade (G3)")

plt.show()
```

```python
df.groupby("studytime")["G3"].agg(["count", "mean", "median"])
```

```python
plt.figure(figsize=(9,6))

sns.scatterplot(data=df, x="absences", y="G3")

plt.title("Absences vs Final Mathematics Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade (G3)")

plt.show()
```

df[["absences", "G3"]].corr()

```python
plt.figure(figsize=(9,6))

sns.boxplot(x="failures", y="G3", data=df)

plt.title("Previous Failures vs Final Mathematics Grade")
plt.xlabel("Number of Previous Failures")
plt.ylabel("Final Grade (G3)")

plt.show()
```

```python
df.groupby("failures")["G3"].agg(["count", "mean", "median"])
```

```python
plt.figure(figsize=(8,6))

sns.boxplot(x="sex", y="G3", data=df)

plt.title("Sex vs Final Mathematics Grade")
plt.xlabel("Sex")
plt.ylabel("Final Grade (G3)")

plt.show()
```

```python
df.groupby("sex")["G3"].agg(["count", "mean", "median"])
```

```python
plt.figure(figsize=(9,6))

sns.boxplot(x="age", y="G3", data=df)

plt.title("Age vs Final Mathematics Grade")
plt.xlabel("Age")
plt.ylabel("Final Grade (G3)")

plt.show()
```

```python
plt.figure(figsize=(9,6))

sns.boxplot(x="Medu", y="G3", data=df)

plt.title("Mother's Education vs Final Mathematics Grade")
plt.xlabel("Mother's Education Level")
plt.ylabel("Final Grade (G3)")

plt.show()
```

```python
plt.figure(figsize=(9,6))

sns.boxplot(x="Fedu", y="G3", data=df)

plt.title("Father's Education vs Final Mathematics Grade")
plt.xlabel("Father's Education Level")
plt.ylabel("Final Grade (G3)")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.boxplot(x="schoolsup", y="G3", data=df)

plt.title("School Support vs Final Mathematics Grade")
plt.xlabel("School Support")
plt.ylabel("Final Grade (G3)")

plt.show()
```

```python
df.groupby("schoolsup")["G3"].agg(["count", "mean", "median"])
```

```python
numeric_df = df.select_dtypes(include=np.number)

correlation = numeric_df.corr()

plt.figure(figsize=(14,10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix of Numerical Variables")

plt.show()
```
